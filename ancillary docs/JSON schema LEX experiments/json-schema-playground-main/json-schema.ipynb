```python
from jsonschema import validate, validators
import json
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
from pathlib import Path
import os
```

```python
with open('./gql-meta-schema', 'r', encoding='utf-8') as gm:
    gql_meta_schema = json.load(gm)

with open('./gql-defs', 'r', encoding='utf-8') as gd:
    gql_defs = json.load(gd)

with open('./database-dialect-database-type', 'r', encoding='utf-8') as ddt:
    dialect = json.load(ddt)

with open('./customer-schema', 'r', encoding='utf-8') as cs:
    customer_schema = json.load(cs)

# with open('./customer-instance.json', 'r', encoding='utf-8') as ci:
#     customer_instance = json.load(ci)
```

```python
# one way: use the schema_store approach
schema_store = {customer_schema['$id']: customer_schema, gql_defs['$id']: gql_defs, gql_meta_schema['$id']: gql_meta_schema, dialect['$id']: dialect}
resolver = validators.RefResolver.from_schema(customer_schema, store=schema_store)
validator = validators.Draft202012Validator(customer_schema, resolver=resolver)
validator.META_SCHEMA = gql_meta_schema
validator.check_schema(customer_schema)

def valid(i):
    if not validator.is_valid(i):
        # display the reason
        validator.validate(i)
    else:
        return True
```
